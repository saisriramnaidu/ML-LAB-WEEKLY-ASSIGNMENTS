**K - MEANS CLUSTERING** IS AN UNSUPERVISED ML ALGORITHM USED TO DIVIDE THE DATASET INTO K DIFFERENT GROUPS BASED ON SIMILARITY DATQA POINTS

```python
#IMPORTING THE LIBRARIES AND DATASETS


import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
data = np.array([2, 4, 10, 12, 3, 20, 30, 11, 25])
M1 = 4
M2 = 11
print("Dataset:", data)
print("Initial Centroid M1:", M1)
print("Initial Centroid M2:", M2)
```

TO FINF=D THE INITIAL DISTANCE OF THE MATRIC **(ITERATION - 1)**:

```python

M1 = 4
M2 = 11
d1 = np.abs(data - M1)
d2 = np.abs(data - M2)
cluster = np.where(d1 <= d2, 1, 2)
matrix1 = pd.DataFrame({
    'Data Points': data,
    'M1 Distance': d1,
    'M2 Distance': d2,
    'Cluster': ['C1' if c == 1 else 'C2' for c in cluster]
})
print(matrix1)
C1 = data[cluster == 1]
C2 = data[cluster == 2]
print("C1:", C1)
print("C2:", C2)
M1 = np.mean(C1)
M2 = np.mean(C2)
print("New M1:", M1)
print("New M2:", M2)
```

THEN WE CAN FIND THE 2ND , 3RD AND 4TH ITERATIONS FOR THE FOLLOWING MATRIX. LATER O TO COMPUTE AUTOMATIC **"K - mEANS CLASSIFICATION"**:

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
data = np.array([2, 4, 10, 12, 3, 20, 30, 11, 25])
M1 = 4
M2 = 11
old_cluster = None
iteration = 1
history = []
while True:
    d1 = np.abs(data - M1)
    d2 = np.abs(data - M2)
    cluster = np.where(d1 <= d2, 1, 2)
    matrix = pd.DataFrame({
        'Data': data,
        'M1 Distance': d1,
        'M2 Distance': d2,
        'Cluster': ['C1' if c == 1 else 'C2' for c in cluster]
    })
    print("\nITERATION", iteration)
    print("M1 =", M1, "M2 =", M2)
    print(matrix.to_string(index=False))
    C1 = data[cluster == 1]
    C2 = data[cluster == 2]
    print("C1 =", C1)
    print("C2 =", C2)
    history.append((iteration, M1, M2))
    if old_cluster is not None and np.array_equal(cluster, old_cluster):
        print("\nClusters are unchanged.")
        print("K-Means Converged!")
        break
    old_cluster = cluster.copy()
    M1 = np.mean(C1)
    M2 = np.mean(C2)
    print("New M1 =", M1)
    print("New M2 =", M2)
    iteration += 1
print("\nFINAL CLASSIFICATION")
print("C1:", C1)
print("C2:", C2)
print("Final Centroids:", M1, M2)
```

TO GET THE **"FINAL K - mEANS CLUSTERING"** THE GRAPH IS DISTRIBUTED B/Q THE POINTS C1 AND C2:

```python

plt.figure(figsize=(9, 4))
plt.scatter(C1, np.zeros(len(C1)), color='blue',
            label='Cluster C1', s=100)
plt.scatter(C2, np.zeros(len(C2)), color='red',
            label='Cluster C2', s=100)
plt.scatter([M1, M2], [0, 0], color='black',
            marker='X', s=200, label='Centroids')
for x in data:
    plt.text(x, 0.04, str(x), ha='center')
plt.title("K-Means Clustering - Final Result")
plt.xlabel("Data Points")
plt.yticks([])
plt.legend()
plt.grid()
plt.show()
```
