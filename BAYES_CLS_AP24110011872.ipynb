```python
#CH. SAI SRIRAM NAIDU  - AP24110011872
#BAYES THEOREM

#Bayes' Theorem – Email Spam Detection
#Suppose a Machine Learning model needs to determine whether an email
#is spam based on the presence of the word "free".
#Assume that:
#1. 20% of all emails are spam.
#2. 80% of spam emails contain the word "free".
#3. 10% of non-spam emails contain the word "free

P_S = 0.20
P_not_S = 0.80

P_F_given_S = 0.80
P_F_given_not_S = 0.10

P_F = (P_F_given_S * P_S) + (P_F_given_not_S * P_not_S)

P_S_given_F = (P_F_given_S * P_S) / P_F

print("Bayes' Theorem - Email Spam Detection")
print("P(Spam) =", P_S)
print("P(Not Spam) =", P_not_S)
print("P(Free | Spam) =", P_F_given_S)
print("P(Free | Not Spam) =", P_F_given_not_S)

print("\nTotal Probability =", P_F)
print("P(Spam | Free) =", round(P_S_given_F, 4))
print("Final Spam Probability =", round(P_S_given_F * 100, 2), "%")
```
