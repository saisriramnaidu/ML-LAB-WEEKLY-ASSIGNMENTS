```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# Load dataset
df = pd.read_csv("/content/WineQT.csv")

# Select features and exclude target and ID
X = df.drop(columns=["quality", "Id"])

# Standardize the features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Apply PCA
pca = PCA()
X_pca = pca.fit_transform(X_scaled)

# Explained variance
explained_variance = pca.explained_variance_ratio_ * 100
cumulative_variance = explained_variance.cumsum()

# Display results
results = pd.DataFrame({
    "Principal Component": [
        f"PC{i+1}" for i in range(len(explained_variance))
    ],
    "Explained Variance (%)": explained_variance,
    "Cumulative Variance (%)": cumulative_variance
})

print(results.round(2))

# Plot explained variance
plt.figure(figsize=(8, 5))
plt.plot(
    range(1, len(explained_variance) + 1),
    cumulative_variance,
    marker="o"
)
plt.axhline(y=90, linestyle="--", label="90% variance")
plt.xlabel("Number of Principal Components")
plt.ylabel("Cumulative Explained Variance (%)")
plt.title("PCA - Wine Quality Dataset")
plt.xticks(range(1, len(explained_variance) + 1))
plt.grid(True)
plt.legend()
plt.show()

# Reduce the dataset to 2 principal components
pca_2 = PCA(n_components=2)
X_2d = pca_2.fit_transform(X_scaled)

pca_df = pd.DataFrame(X_2d, columns=["PC1", "PC2"])
print("\nFirst five transformed records:")
print(pca_df.head())

print("\nVariance retained by 2 components:",
      round(pca_2.explained_variance_ratio_.sum() * 100, 2), "%")
```

Question 3: What happens if we retain more components?
Try 1, 2, 3, 5 and 10 principal components.
• How does the cumulative explained variance change?
ANS) Cumulative variance increases as we add more components. In our output, 2 components retain 45.77%, 3 retain 60.07%, and 5 retain 79.80%.
• How many components are needed to retain at least 90% of the variance?
ANS) We need 7 components, which retain 90.92% of the total variance.
• Is reducing to two components always the best choice?
ANS) No, because 2 components retain only 45.77% of the variance. More components preserve more information.


```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
import numpy as np

# Load dataset
df = pd.read_csv("/content/WineQT.csv")

# Select features and exclude target and ID
X = df.drop(columns=["quality", "Id"])

# Standardize the features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Try different numbers of components
for n in [1, 2, 3, 5, 10]:
    pca = PCA(n_components=n)
    X_pca = pca.fit_transform(X_scaled)

    variance = pca.explained_variance_ratio_.sum() * 100

    print(f"{n} components retain {variance:.2f}% variance")

# Find components needed to retain at least 90% variance
pca = PCA()
pca.fit(X_scaled)

cumulative_variance = np.cumsum(pca.explained_variance_ratio_)
n_components = np.argmax(cumulative_variance >= 0.90) + 1

print("\nComponents needed for 90% variance:", n_components)
print("Variance retained:", round(cumulative_variance[n_components - 1] * 100, 2), "%")
```

Question 4: What happens to the visualization?
Plot the first two principal components and observe the three wine classes.
• Are the classes clearly separated?
ANS) The classes may overlap when plotted using the first two components. PCA does not guarantee clear separation between classes.
• What information might be lost when reducing 13 features to 2?
ANS) Some information from the original features is lost. In our output, the first two components retain only 45.77% of the variance.
• Does PCA use the class labels to create the components?
ANS) No, PCA does not use class labels. It uses only the input features to create the principal components.


```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# Load dataset
df = pd.read_csv("/content/WineQT.csv")

# Select features and exclude ID and target
X = df.drop(columns=["quality", "Id"])
y = df["quality"]

# Standardize the features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Apply PCA with 2 components
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

# Create a DataFrame
pca_df = pd.DataFrame(X_pca, columns=["PC1", "PC2"])
pca_df["quality"] = y.values

# Plot the first two principal components
plt.figure(figsize=(8, 5))

for wine_class in sorted(pca_df["quality"].unique()):
    data = pca_df[pca_df["quality"] == wine_class]
    plt.scatter(data["PC1"], data["PC2"], label=f"Quality {wine_class}")

plt.xlabel("Principal Component 1 (PC1)")
plt.ylabel("Principal Component 2 (PC2)")
plt.title("PCA - Wine Quality Classes")
plt.legend(title="Wine Quality")
plt.grid(True)
plt.show()

# Display variance retained
print("Variance retained by 2 components:",
      round(pca.explained_variance_ratio_.sum() * 100, 2), "%")

print("\nDoes PCA use class labels to create components? No.")
```

Question 5: Does PCA select existing features?
• Are the two principal components original Wine dataset columns?
ANS) No, PC1 and PC2 are new features created by combining information from the original features
• What is the difference between feature selection and feature extraction?
ANS) Feature selection chooses existing features from the dataset, while feature extraction creates new features from the original ones.
• Why is PCA considered a feature extraction technique?
ANS) PCA creates new features called principal components by combining the original features. It reduces the dataset while retaining as much information as possible.


```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# Load dataset
df = pd.read_csv("/content/WineQT.csv")

# Select features and exclude ID and target
X = df.drop(columns=["quality", "Id"])

# Standardize the features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Apply PCA with 2 components
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

# Create a new dataset with 2 principal components
pca_df = pd.DataFrame(X_pca, columns=["PC1", "PC2"])

# Display original features
print("Original Features:")
print(X.columns.tolist())

# Display new principal components
print("\nNew Principal Components:")
print(pca_df.head())

# Compare the number of features
print("\nOriginal Number of Features:", X.shape[1])
print("Features After PCA:", pca_df.shape[1])

# Display explained variance
print("\nVariance Retained:",
      round(pca.explained_variance_ratio_.sum() * 100, 2), "%")

print("\nConclusion:")
print("PC1 and PC2 are new features, not original columns.")
print("PCA is a feature extraction technique.")
```
