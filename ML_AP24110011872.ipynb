## **DECISION TREE**:

WITH A DATA SET WE ARE MAKING A ENTROPY, INFORMATION GAIN AND DECISION TREE.

**ENTROPY**:

ENTROPY(s) = - p1 * log_2 (p1) - p0 * log_2 (p0)

**INFORMATION GAIN**:


INFORMATION GAIN = ENTROPY (S) - ENTROPY (S / A) -> ENTROPY PER PARTICULAR ATTRIBUTE.

```python
import math
import pandas as pd
data = [
    ["D1", "Sunny",    "Hot",  "High",   "Weak",   "No"],
    ["D2", "Sunny",    "Hot",  "High",   "Strong", "No"],
    ["D3", "Overcast", "Hot",  "High",   "Weak",   "Yes"],
    ["D4", "Rain",     "Mild", "High",   "Weak",   "Yes"],
    ["D5", "Rain",     "Cool", "Normal", "Weak",   "Yes"],
    ["D6", "Rain",     "Cool", "Normal", "Strong", "No"],
    ["D7", "Overcast", "Cool", "Normal", "Strong", "Yes"],
    ["D8", "Sunny",    "Mid",  "High",   "Weak",   "No"],
    ["D9", "Sunny",    "Cool", "Normal", "Weak",   "Yes"],
    ["D10","Rain",     "Mid",  "Normal", "Weak",   "Yes"],
    ["D11","Sunny",    "Mid",  "Normal", "Strong", "Yes"],
    ["D12","Overcast", "Mid",  "High",   "Strong", "Yes"],
    ["D13","Overcast", "Hot",  "Normal", "Weak",   "Yes"],
    ["D14","Rain",     "Mid",  "High",   "Strong", "No"]
]
df = pd.DataFrame(data, columns=["Day", "Outlook", "Temperature","Humidity", "Wind", "PlayTennis"])
print(df)
```

```python
#TO CHECK THE COUNT OF YES AND NO
print("\nClass Counts:")
print(df["PlayTennis"].value_counts())
```

```python
#FOR CREATING THE ENTROPY FUNCTION FORMULA
def entropy(data):
    total = len(data)
    counts = data.value_counts()
    ent = 0
    for count in counts:
        probability = count / total
        ent -= probability * math.log2(probability)
    return ent
```

```python
#FOR OVERALL ENTROPY
overall_entropy = entropy(df["PlayTennis"])
print("Overall Entropy =", round(overall_entropy, 4))
```

```python
#CALCULATE THE ENTROPY OF ATTRIBUTE
def attribute_entropy(df, attribute):
    total = len(df)
    weighted_entropy = 0
    for value in df[attribute].unique():
        subset = df[df[attribute] == value]
        e = entropy(subset["PlayTennis"])
        weight = len(subset) / total
        weighted_entropy += weight * e
        print(attribute, "=", value, "Count =", len(subset),"Entropy =", round(e, 4))
    return weighted_entropy
```
