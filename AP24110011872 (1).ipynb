
Step 4: Comparing Two Sentences (Cosine Similarity)

Once text is turned into number vectors, we can measure how similar two sentences are using cosine similarity — a formula you'll see constantly in NLP and recommendation systems:

cosine similarity=A⋅B∥A∥×∥B∥

- This measures the angle between two vectors, not their length — so it works even if one sentence is longer than another.

```python
import numpy as np

# Sentences
sentence1 = "I like rain"
sentence2 = "I enjoy rain"

sentence3 = "I like rain"
sentence4 = "The weather is cloudy today"

def bow(sentence, vocabulary):
    words = sentence.split()      # to create Bag-of-Words vector
    return np.array([words.count(word) for word in vocabulary])

def cosine_similarity(A, B):  # Function used to calculate cosine similarity
    return np.dot(A, B) / (np.linalg.norm(A) * np.linalg.norm(B))

vocab1 = np.unique(sentence1.split() + sentence2.split())   #USING FOR SIMILAR STATEMENTS   AP24110011872
A = bow(sentence1, vocab1)
B = bow(sentence2, vocab1)

similarity1 = cosine_similarity(A, B)

print("Similar pair similarity:", similarity1)


vocab2 = np.unique(sentence3.split() + sentence4.split())   # USING FOR DIFFERENT STATEMENT AP24110011872
C = bow(sentence3, vocab2)
D = bow(sentence4, vocab2)

similarity2 = cosine_similarity(C, D)

print("Different pair similarity:", similarity2)    #AP24110011872
```

Take any paragraph of your choice (3-4 sentences, write it below as a string). Then: (AP24110011872)

Use NumPy to find and print the top 3 most frequent words (hint: look at sorting — np.argsort() sorts indices by value, and slicing with [::-1] reverses order).

```python
import numpy as np
paragraph = """
Mobile editing has become very popular among content creators.
Mobile editing apps make it easy to edit photos and videos anywhere.
Good mobile editing can improve the quality of social media content.
Many creators use mobile editing to add music, effects, text, and transitions.
"""
tokens = paragraph.lower().split()    # Tokenizing the paragraph
vocabulary = np.unique(tokens)      # AP24110011872 Build vocabulary

# Build word-count vector
word_counts = np.array([tokens.count(word) for word in vocabulary])

print("Vocabulary:", vocabulary)
print("Word counts:", word_counts)

# 3. Find the top 3 most frequent words
sorted_indices = np.argsort(word_counts)[::-1]

top_3 = sorted_indices[:3]

print("\nTop 3 most frequent words:")

for i in top_3:
    print(vocabulary[i], ":", word_counts[i])
```

```python
import numpy as np
sentence1 = "I love editing videos on my mobile phone"    # CREATING 3 SENTENCES AP24110011872
sentence2 = "I enjoy editing videos on my smartphone"
sentence3 = "I like playing football with my friends"

tokens1 = sentence1.lower().split()   #TOKENIZING THOSE 3 SENTENCES AP24110011872
tokens2 = sentence2.lower().split()
tokens3 = sentence3.lower().split()

all_tokens = tokens1 + tokens2 + tokens3    #BUILD A SINGLE VOCABULARY SHARED TO ALL 3
vocabulary = np.unique(all_tokens)

print("Vocabulary:")
print(vocabulary)

def bag_of_words_vector(tokens, vocabulary):      #CREAING A BAG OF WORDS VECTORS AP24110011872
    return np.array([tokens.count(word) for word in vocabulary])

vector1 = bag_of_words_vector(tokens1, vocabulary)
vector2 = bag_of_words_vector(tokens2, vocabulary)
vector3 = bag_of_words_vector(tokens3, vocabulary)


print("\nVector 1:", vector1)
print("Vector 2:", vector2)
print("Vector 3:", vector3)

def cosine_similarity(A, B):    #COSINE SIMILARITY FUCN AP24110011872
    return np.dot(A, B) / (np.linalg.norm(A) * np.linalg.norm(B))

sim12 = cosine_similarity(vector1, vector2)   # Calculating the 3 pairs
sim13 = cosine_similarity(vector1, vector3)
sim23 = cosine_similarity(vector2, vector3)
print("\nSimilarity 1-2:", sim12)
print("Similarity 1-3:", sim13)
print("Similarity 2-3:", sim23)

similarities = {          #finiding the most similarity pair ap24110011872
    "Sentence 1 and Sentence 2": sim12,
    "Sentence 1 and Sentence 3": sim13,
    "Sentence 2 and Sentence 3": sim23
}

most_similar = max(similarities, key=similarities.get)
print("\nMost similar pair:", most_similar)
print("Similarity score:", similarities[most_similar])
```
